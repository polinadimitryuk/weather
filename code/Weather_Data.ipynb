#### Define a column that labels each day by day of the year: 1, 2, 3, ..., 365

```python
df['day_of_year'] = pd.DatetimeIndex(df['date']).day_of_year
```

```python
df.head()
```

#### Compute the mean precipitation for each day in Seattle, averaged across years. 

```python
mean_day_precipitation_sea=df.loc[
    df['city']=='SEA',
    ['precipitation','day_of_year']
].groupby(
    'day_of_year'
).mean()
```

##### Compute the mean precipitation for each day in Pittsburgh

```python
mean_day_precipitation_pb=df.loc[
    df['city']=='PIT',
    ['precipitation','day_of_year']
].groupby(
    'day_of_year'
).mean()
```

```python
mean_day_precipitation_sea
```

```python
mean_day_precipitation_pb
```

```python
plt.figure(figsize=(20,5))

sns.lineplot(data=mean_day_precipitation_sea, x='day_of_year', y='precipitation', label='Seattle')
sns.lineplot(data=mean_day_precipitation_pb, x='day_of_year', y='precipitation', label='Pittsburgh')

plt.xlabel('Day of the year', fontsize=18)
plt.ylabel('Mean precipitation (inches)', fontsize=18)

plt.tick_params(labelsize=15)

plt.show()
```

#### Get the index of each row where precipitation is missing.

```python
indices = np.where(df['precipitation'].isna() == True)[0]
```

```python
indices
```

```python
len(indices)
```

#### Replace each missing value with the mean on that day

We will make a loop, and if the row for Seattle (SEA), we will take the mean precipitation for that day of the year and will insert to 
the row with missing data. For Pittburgh (PIT), we will use Pittburgh's mean precipitation for that day of the year.

```python
for index in indices:
    if df.loc[index,'city']=='SEA': df.loc[index,'precipitation'] = mean_day_precipitation_sea.loc[df.loc[index, 'day_of_year']].values[0]
    elif df.loc[index,'city']=='PIT': df.loc[index,'precipitation'] = mean_day_precipitation_pb.loc[df.loc[index, 'day_of_year']].values[0]
```
