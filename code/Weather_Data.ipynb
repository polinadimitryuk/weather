#### Get the index of each row where precipitation is missing.

```python
indices = np.where(df['precipitation'].isna() == True)[0]
```

```python
indices
```

```python
len(indices)
```

#### Replace each missing value with the mean on that day

We will make a loop, and if the row for Seattle (SEA), we will take the mean precipitation for that day of the year and will insert to 
the row with missing data. For Pittburgh (PIT), we will use Pittburgh's mean precipitation for that day of the year.

```python
for index in indices:
    if df.loc[index,'city']=='SEA': df.loc[index,'precipitation'] = mean_day_precipitation_sea.loc[df.loc[index, 'day_of_year']].values[0]
    elif df.loc[index,'city']=='PIT': df.loc[index,'precipitation'] = mean_day_precipitation_pb.loc[df.loc[index, 'day_of_year']].values[0]
```

### Analysis

#### Perform a statistical test for differences in the mean precipitation each month between cities

```python
plt.figure(figsize=(5, 4))

sns.histplot(data=df.loc[df['month'] == 1], x='precipitation', hue='city')

plt.xlabel('Precipitation (inches)', fontsize=15)
plt.ylabel('Number of days', fontsize=15)
plt.title('Precipitation in January', fontsize=15)

plt.tick_params(labelsize=15)

plt.show()
```

```python
from scipy import stats
```

```python
significance_level = 0.05
significantly_different = np.zeros(12)

# Perform t-test for each month
for month in range(1, 13):
    # Get precipitation data for Seattle and Pittsburgh for current month
    sea_data = df.loc[(df['city'] == 'SEA') & (df['month'] == month), 'precipitation']
    pb_data = df.loc[(df['city'] == 'PIT') & (df['month'] == month), 'precipitation']

    t_statistic, p_value = stats.ttest_ind(sea_data, pb_data, equal_var=False)

    if p_value < significance_level:
        significantly_different[month-1] = 1
        
    print(f"Month {month}:")
    print(f" t-statistic={t_statistic:.2f}")
    print(f" p-value t test={p_value:.3f}")
    print("-" * 20)
```

Mean daily precipitation is significantly higher in Seattle in January, November, and December. 
Mean daily precipitation is significantly higher in Pittsburgh in April, May, July, and August. 
No statistically significant difference was detected in February, March, June, September, October.

#### Plot the mean Precipitation each month with a star for significant differences

```python
plt.figure(figsize=(20, 4))

sns.barplot(data=df, x='month', y='precipitation', hue='city', alpha=0.75)

plt.xlabel('Month', fontsize=15)
plt.ylabel('Precipitation (inches)', fontsize=15)
plt.title('Mean monthly precipitation', fontsize=15)

plt.tick_params(labelsize=15)

plt.xticks(ticks=range(12), labels=month_names) 

# Add stars for significantly different months
for month in range(12):
    if significantly_different[month] ==1:

        # Add a star
        plt.text(month, 0.3, '*', ha='center', fontsize=25)
        
plt.show()
```

#### Perform a statistical test for differences in the proportion of days with any precipitation each month between the cities

```python
from statsmodels.stats.proportion import proportions_ztest

significance_level = 0.05
significantly_different_proportion = np.zeros(12)

#Perform t-test for each month
for month in range(1, 13):

    # Create a contingency table for Seattle and Pittsburgh for current month:
    contingency_table = pd.crosstab(
        df.loc[df['month'] == month, 'city'], df.loc[df['month'] == month, 'any_precipitation']
    )

    # Calculate the number of True values (days with precipitation) for each city
    days_with_precipitation = contingency_table[True]

    # Calculate the total number of days for each city
    total_counts = contingency_table.sum(axis=1)

    # Hypothesis test
    zstat, p_value = proportions_ztest(
    count=days_with_precipitation, nobs=total_counts, alternative='two-sided')

    if p_value < significance_level:
        significantly_different_proportion[month-1] = 1
        
    print(f"Month {month}:")
    print(f" z-statistic={zstat:.2f}")
    print(f" p-value ={p_value:.3f}")
    print("-" * 20)
```
